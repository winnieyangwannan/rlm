# Load

```python
import glob
import os

run_id = 513
model_name = "gpt-5"
job_name = "comparison"
task_name = "freesound-audio-tagging-2019"
log_dir = "/checkpoint/maui_sft/winnieyangwn/rlm_dumps"
codebase_extensions = [".jsonl"]

# Option 1: Provide a specific log name to load directly
log_name = None  # e.g., "gpt-5_comparison_513_20260208_123456.jsonl"

if log_name:
    # Directly load the specified log file
    LOG_PATH = os.path.join(log_dir, log_name)
    if not os.path.exists(LOG_PATH):
        raise FileNotFoundError(f"Specified log file not found: {LOG_PATH}")
    print(f"Loading specified log: {LOG_PATH}")
else:
    # Use underscore after run_id to avoid matching "summarization_comparison_*" logs
    LOG_PATH_PREFIX = f"{log_dir}/{model_name}_{job_name}_{run_id}_{task_name}"

    # Find all log files matching the prefix pattern (only .jsonl files)
    matching_logs = glob.glob(f"{LOG_PATH_PREFIX}*.jsonl")

    if not matching_logs:
        raise FileNotFoundError(f"No log files found matching prefix: {LOG_PATH_PREFIX}*.jsonl")

    # Get the most recent log file by modification time
    LOG_PATH = max(matching_logs, key=os.path.getmtime)
    print(f"Found {len(matching_logs)} matching log file(s)")
    print(f"Loading most recent: {LOG_PATH}")

# Load the log - first entry is metadata, rest are iterations
entries = load_rlm_log(LOG_PATH)
metadata = entries[0]
iterations = entries[1:]

print(f"Loaded {len(iterations)} iterations")
```

```python
import shutil

def archive_file(file_path: str, archive_dir: str = "/checkpoint/maui_sft/winnieyangwn/rlm_dumps/archive/gpt5/comparison/") -> str:
    """Move a file to the archive directory.
    
    Args:
        file_path: Path to the file to archive
        archive_dir: Destination archive directory
        
    Returns:
        New path of the archived file
    """
    dest_path = shutil.move(file_path, archive_dir)
    print(f"Archived: {file_path} -> {dest_path}")
    return dest_path

```
