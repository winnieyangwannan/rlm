# Extract Final Answer and save

```python
import json
import ast
import re

# Parse final_answer string to list of dicts
# Use ast.literal_eval since it uses single quotes (Python format, not JSON)
data = ast.literal_eval(final_answer)

# Extract datetime and hash from original log filename to maintain same structure
# Pattern: _YYYY-MM-DD_HH-MM-SS_HASH.jsonl
log_basename = os.path.basename(LOG_PATH)
datetime_match = re.search(r'_(\d{4}-\d{2}-\d{2}_\d{2}-\d{2}-\d{2}_[a-f0-9]+)\.jsonl$', log_basename)

if datetime_match:
    datetime_suffix = datetime_match.group(1)
    output_filename = f"{job_name}_extracted_{datetime_suffix}.jsonl"
else:
    # Fallback if pattern not found
    output_filename = f"{job_name}_extracted.jsonl"

output_path = os.path.join(log_dir, output_filename)

# Save as JSONL (one JSON object per line)
with open(output_path, 'w') as f:
    for item in data:
        f.write(json.dumps(item) + '\n')

print(f"Saved {len(data)} items to: {output_path}")
```

```python
import pandas as pd

# Load the JSONL file
df = pd.read_json(output_path, lines=True)

print(f"Loaded {len(df)} rows from: {output_path}")
print(f"Columns: {df.columns.tolist()}")

```
